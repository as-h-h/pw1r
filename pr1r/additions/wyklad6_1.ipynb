### Unicode

Unicode - międzynarodowy standard kodowania symboli ("wszystkich" alfabetów i kwazialfabetów, np. symboli drogowych, emojis). Obejmuje ~150000 znaków (stan na 2026 r.).

* [home.unicode.org](https://home.unicode.org)
* [Strony kodowe](https://www.unicode.org/charts/) (np. [alfabet grecki](https://www.unicode.org/charts/PDF/U0370.pdf), [emotikony](https://www.unicode.org/charts/PDF/U1F600.pdf), [operatory matematyczne](https://www.unicode.org/charts/PDF/U2200.pdf))

Każdy symbol ma przypisaną liczbę - punkt kodowy (zazwyczaj pisaną w systemie heksadecymalnym, tj. szesnastkowym). Napisy w Pythonie mogą składać się z dowolnych symboli Unicode (wewnętrznie napis to ciąg punktów kodowych). Podobnie wyglądają napisy w wielu współczesnych językach programowania.

Wstawianie symboli Unicode do napisów: `\xXX`, `\uXXXX` lub `\UXXXXXXXX` (gdzie `X` to cyfra heksadecymalna). Przykłady:

`isdecimal()`, `isdigit()`, `isnumeric()` - cyfra dziesiętna, cyfra, numerał

```python
'1234'.isdecimal()
```

```python
'1234❸'.isdecimal()
```

```python
'1234❸'.isdigit()
```

Ogólnie: `isdecimal()` $\implies$ `isdigit()` $\implies$ `isnumeric()`.

To, czy dla znaku spełnione jest "bycie cyfrą" etc. jest zdefiniowane w Unicode.

```python
s = "12१२३४⓿❶❷❸¾⅖⅗⅘⒑⒒፼"

print(f'{"char":^8}{"isdecimal?":^12}{"isdigit?":^12}{"isnumeric?":^12}')
for c in s:
    print(f'{c:^8}{c.isdecimal():^12}{c.isdigit():^12}{c.isnumeric():^12}')
```

### Formatowanie

`upper()`, `lower()`, `casefold()`

```python
"abcdEFG".upper()
```

```python
"abcdEFG".lower()
```

```python
"α".upper()
```

```python
"α".upper() == "a".upper()
```

```python
"α".upper().isupper()
```

```python
s = "ß"
print(s.upper())
print(s.lower())        
print(s.casefold())
```

`casefold()` jest przydatny przy porównywaniu tych samych słów z alternatywną pisownią znaków:

```python
"groß".lower() == "gross".lower()
```

```python
"groß".casefold() == "gross".casefold()
```

Podobny przykład: σ i ς - dwie małe litery sigma.
