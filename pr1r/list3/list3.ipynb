```python
def binom(n, k):
    if k == 0 or n == k:
        return 1
    return binom(n-1, k) + binom(n-1, k-1)
```

```python
print(binom(5, 2))
```

```python
import time
t_0 = time.perf_counter()
binom(30, 15)
t_1 = time.perf_counter()
print(f"Czas: {t_1 - t_0} sekund")
```

```python
def pascal(n):
    triangle = [[1]]
    for i in range(n):
        prev_row = triangle[-1]
        cur_row = [1] + [prev_row[j] + prev_row[j + 1] for j in range(i)] + [1]
        triangle.append(cur_row)
    return triangle
```

```python
t_0 = time.perf_counter()
pascal(30)
t_1 = time.perf_counter()
print(f"Czas: {t_1 - t_0:2.15f} sekund")
```

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

```python
print(fibonacci(7))
```
