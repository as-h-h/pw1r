```python
class Rectangle:
    def __init__(self, width=1.0, height=1.0):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def scale(self, factor=1.0):
        if factor < 0:
            raise ValueError("Scale factor must be positive.")
        return Rectangle(self.width * factor, self.height * factor)
```

```python
from math import gcd

class Fraction:
    def __init__(self, numer=0, denom=1):
        comm = gcd(numer, denom)
        self.numer = numer // comm
        self.denom = denom // comm
        self.fraction = numer/denom

    def __abs__(self):
        return Fraction(abs(self.numer), abs(self.denom))
    
    def __neg__(self):
        return Fraction(-self.numer, self.denom)

    def __str__(self):
        return f"{self.numer}/{self.denom}"

    def __add__(self, other_frac):
        return Fraction(self.numer*other_frac.denom + other_frac.numer*self.denom, self.denom * other_frac.denom)

    def __sub__(self, other_frac):
        return Fraction(self.numer*other_frac.denom - other_frac.numer*self.denom, self.denom * other_frac.denom)

    def __mul__(self, other_frac):
        return Fraction(self.numer*other_frac.numer, self.denom*other_frac.denom)

    def __truediv__(self, other_frac):
        return Fraction(self.numer*other_frac.denom, self.denom*other_frac.numer)
```

```python
import numpy as np
import matplotlib.pyplot as plt

class TurtlePath:
    def __init__(self):
        self.positions = np.array([[0.0, 0.0]])
        self.direction = 0

    def left(self, degrees):
        self.direction += degrees

    def right(self, degrees):
        self.direction -= degrees

    def forward(self, distance):
        last_x, last_y = self.positions[-1]
        x = last_x + distance * np.cos(np.radians(self.direction))
        y = last_y + distance * np.sin(np.radians(self.direction))
        self.positions = np.vstack([self.positions, [[x, y]]])

    def draw(self, color="black"):
        plt.plot(self.positions[:, 0], self.positions[:, 1], color=color)
```

```python
class Board:  
    def __init__(self, board_file, treasure_file):
        self.rooms = self.adjacent_rooms(board_file)
        self.treasure = self.treasure_rooms(treasure_file)

    def adjacent_rooms(self, input_file):
        d = {}
        with open(input_file, encoding='utf8') as f:
            for line in f:
                rooms = line.strip().split(':')
                d[rooms[0]] = rooms[1].split(',')
        return d

    def treasure_rooms(self, input_file):
        d = {}
        with open(input_file, encoding='utf8') as f:
            for line in f:
                rooms = line.strip().split(',')
                d[rooms[0]] = int(rooms[1])
        return d
    
    def num_close_treasures(self, chamber):
        num = 0
        adjacent_chambers = self.rooms[chamber]
        for chambs in adjacent_chambers:
            num += self.treasure[chambs]
        return num
    
    def get_nearby_treasures(self, chamber):
        lst = []
        adjacent_chambers = self.rooms[chamber]
        for chambs in adjacent_chambers:
            if self.treasure[chambs] == 1:
                lst.append(chambs)
        if len(lst) == 0:
            return "Brak skarbu w pobliżu"
        return lst

    def competing_chambers(self, chamber1, chamber2):
        chamb1_neighboors = self.get_nearby_treasures(chamber1)
        chamb2_neighboors = self.get_nearby_treasures(chamber2)
        neighbours = list(set(chamb1_neighboors) & set(chamb2_neighboors))
        for neighbour in neighbours:
            if self.treasure[neighbour] == 1:
                return True
        return False
```

```python
board = Board("board.txt", "treasures.csv")
print(board.num_close_treasures("Hol"))
print(board.get_nearby_treasures("Most zwodzony"))
print(board.competing_chambers("Most zwodzony", "Spichlerz"))
```
