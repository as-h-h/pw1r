```python
def is_magic_square(array):
    n = len(array)
    magic_sum = sum(array[0])

    for i in range(1, n):
        if np.sum(array, axis=0)[i] != magic_sum or np.sum(array, axis=1)[i] != magic_sum:
            return False
        
    if np.sum(np.diag(array)) != magic_sum or np.sum(np.diag(np.fliplr(array))) != magic_sum:
            return False

    if np.unique(array).size != n**2:
            return False

    return True
```

```python
is_magic_square(np.array([[8, 1, 6], [3, 5, 7], [4, 9, 2]]))
```

```python
is_magic_square(np.array([[1, 1], [1, 1]])) 
```

```python
import numpy as np

def check_winner(board):

    for player in (1, 2):
        p = (board == player)

        # pionowo, poziomo, przekątna
        if (
            np.max(p[:-3, :] & p[1:-2, :] & p[2:-1, :] & p[3:, :])
            or np.max(p[:, :-3] & p[:, 1:-2] & p[:, 2:-1] & p[:, 3:])
            or np.max(p[:-3, :-3] & p[1:-2, 1:-2] & p[2:-1, 2:-1] & p[3:, 3:])
            or np.max(p[:-3, 3:] & p[1:-2, 2:-1] & p[2:-1, 1:-2] & p[3:, :-3])
        ):
            return player

    return None
```

```python
check_winner(board)
```

```python
def is_valid_position(board):
    valid_board = (board > 0)
    top = valid_board[:-1, :]
    bottom = valid_board[1:, :]
    result = np.max(top > bottom)

    if result:
        return False
    
    count1 = np.sum(board == 1)
    count2 = np.sum(board == 2)

    if np.abs(count1 - count2) > 1:
        return False

    return True
```

```python
is_valid_position(board)
```
