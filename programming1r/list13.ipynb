```python
import numpy as np
import pandas as pd
from wyklad13_knn import KNNClassifier
from wyklad13_standardscaler import StandardScaler
from wyklad13_train_test_split import train_test_split
```

```python
df = pd.read_csv("wine.txt")
X  = df.drop(columns=["class"]).values
y  = df["class"].values
k = 5
n_repeats = 50
train_size = 0.7

accuracies = []
for _ in range(n_repeats):
    X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size)
    clf = KNNClassifier(k=k)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    accuracies.append(np.mean(y_pred == y_test))

accuracies = np.array(accuracies)
print(f"k={k}: max={accuracies.max():.4f}  min={accuracies.min():.4f}  średnia={accuracies.mean():.4f}")
```

```python
k_max = 30
results = {}

for k in range(1, k_max + 1):
    accs = []
    for _ in range(n_repeats):
        X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=train_size)
        clf = KNNClassifier(k=k)
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        accs.append(np.mean(y_pred == y_test))
    accs = np.array(accs)
    results[k] = (accs.max(), accs.min(), accs.mean())
    #print(f"k={k}: max={accs.max():.4f}  min={accs.min():.4f}  mean={accs.mean():.4f}")

best_k = max(results, key=lambda k: results[k][2])
best_mean = results[best_k][2]
print(f"Najlepsze k = {best_k} średnia skuteczność = {best_mean:.4f})")
```

```python
df_iris = pd.read_csv("iris.csv")

classes_iris = df_iris["class"].unique()
label_map = {c: i for i, c in enumerate(classes_iris)}
df_iris["class"] = df_iris["class"].map(label_map)

X_iris = df_iris.drop(columns=["class"]).values.astype(float)
y_iris = df_iris["class"].values.astype(float)
```

```python
k_iris = best_k
acc_plain  = []
acc_scaled = []

for _ in range(n_repeats):
    X_train, X_test, y_train, y_test = train_test_split(X_iris, y_iris, train_size=train_size)

    clf = KNNClassifier(k=k_iris)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    acc_plain.append(np.mean(y_pred == y_test))

    scaler = StandardScaler()
    X_train_s = scaler.fit_transform(X_train)
    X_test_s = scaler.transform(X_test)

    clf_s = KNNClassifier(k=k_iris)
    clf_s.fit(X_train_s, y_train)
    y_pred_s = clf_s.predict(X_test_s)
    acc_scaled.append(np.mean(y_pred_s == y_test))

mean_plain  = np.mean(acc_plain)
mean_scaled = np.mean(acc_scaled)

print(f"{mean_plain:.4f}")
print(f"{mean_scaled:.4f}")
```
