Zadanie 1

```python
numbers = []

while True:
    entry = input()
    
    if entry == 'end':
        if len(numbers) == 0:
            print("Błąd")
        break
    
    numbers.append(float(entry))


product = 1
for x in numbers:
    product *= x
    
geo_avg = product ** (1 / len(numbers))
print(f'{geo_avg:.4f}')
```

```python
n = 3
M = [[0] * n] * n
print(M)
M[0][0] = 1
print(M)
```

```python
def identity(n):
    matrix = [[0] * n for i in range(n)]
    matrix[0][0] = 1
    return matrix

print(identity(3))
```
