```python
def minimal_age(n):
    d = {}
    for person in data["people"]:
        city_name = person["city"]
        person_info = {"age": person["age"], "name": f'{person["name"]} {person["surname"]}'}

        if city_name not in d:
            d[city_name] = [person_info]
        else:
            d[city_name].append(person_info)

    for city_name, people in d.items():
        min_age = min([person["age"] for person in people])
        youngest_person = [p for p in people if p["age"] == min_age]
        if min_age >= n:
            print(f"{city_name}: Najmłodsza osoba: {youngest_person[0]["name"]} w wieku {min_age} lat.")

minimal_age(30)
```
