Zadanie 3

```python
def how_many_k(n, m):
    count = 0
    for k in range(n, m):
        if k % 10 in [1, 3]:
            count += 1
    return count
```

```python
n = int(input("input n: "))
m = int(input("input m: "))
print(f'ilość takich liczb k: {how_many_k(n, m)}')
```

```python
def count_words(string):
    words = string.split()
    df = pd.DataFrame(words)
    return df.mode()

string = requests.get('https://www.math.uni.wroc.pl/~jagiella/files/p1python_r/lorem.txt').text
count_words(string)
```

```python
def count_words(string):
    words = string.split()
    df = pd.DataFrame(words)
    df = df.groupby(0)
    df = df.size().sort_values(ascending=False)
    return df

string = requests.get('https://www.math.uni.wroc.pl/~jagiella/files/p1python_r/lorem.txt').text
count_words(string)
```

```python
def random_route(n):
    x, y = 0, 0
    route_x = [x]
    route_y = [y]

    directions = [(0,1), (0,-1), (1,0), (-1,0)]
    for _ in range(n):
        step_x, step_y = random.choice(directions)
        x += step_x
        y += step_y
        route_x.append(x)
        route_y.append(y)
    plt.figure(figsize=(7, 7))
    plt.grid()
    plt.plot(route_x, route_y, 'o')
```

```python
random_route(10)
```
