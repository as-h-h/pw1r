```python
import string
import matplotlib.pyplot as plt
import numpy as np
```

Zadanie 1

```python
def filter_lines(input_file, output_file, substr):
    with open(output_file, 'w') as g:
        with open(input_file, 'r') as f:
            for line in f:
                if substr in line:
                    g.write(line)
```

```python
filter_lines('text.txt', 'a.txt', 'dolor')
```

```python
def word_wrap(s, max_length=80):
    output = []
    char_count = 0
    words = s.split()
    for word in words:
        char_count += len(word)
        if char_count > max_length:
            print(" ".join(output))
            char_count = len(word)
            output = []
        output.append(word)
    print(" ".join(output))
        
```

```python
word_wrap("Ala ma kota a kot ma Alę i Szczebrzeszyn z tego słynie.", 6)
```

```python
def list_words_in_file(input_file):
    word_list = []
    with open(input_file, encoding='utf8') as f:
        for line in f:
            words = line.split()
            for word in words:
                word = word.strip(string.punctuation).casefold()
                if word == '':
                    continue
                if word.isnumeric():
                    continue
                if "--" in word:
                    word_list += word.split("--")
                    continue
                word_list.append(word)
    return word_list
```

```python
list_words_in_file("text.txt")
```

```python
def forward(step, direction, positions):
    last_x, last_y = positions[-1]
    x = last_x + step * np.cos(np.radians(direction))
    y = last_y + step * np.sin(np.radians(direction))
    positions = np.vstack([positions, [[x, y]]])
    return positions
```

```python
commands_file = 'commands2.txt'

def turle_path(commands_file):
    positions = np.array([[0.0, 0.0]])
    direction = 0

    with open(commands_file, 'r') as file:
        for line in file:
            command = line.split()

            if command[0] == 'forward':
                positions = forward(float(command[1]), direction, positions)
            elif command[0] == 'turn':
                direction += int(command[1])

    plt.plot(positions[:, 0], positions[:, 1])

turle_path(commands_file)
```
